### Imports

```python
# 01_preprocessing.ipynb
from src.preprocessing import TextPreprocessor
from src.dataset_builder import DatasetBuilder


import os

```

```python
# On récupère le chemin du dossier du notebook
NOTEBOOK_DIR = os.getcwd()

# Le projet est un dossier au-dessus : assurance-llm-assistant/
PROJECT_ROOT = os.path.abspath(os.path.join(NOTEBOOK_DIR, ".."))

# Chemins des dossiers
RAW_DATA_DIR = os.path.join(PROJECT_ROOT, "data", "raw")
OUTPUT_DATASET = os.path.join(PROJECT_ROOT, "data", "processed", "dataset.jsonl")

print("Notebook directory :", NOTEBOOK_DIR)
print("Project root       :", PROJECT_ROOT)
print("RAW exists?        :", os.path.exists(RAW_DATA_DIR))
print("RAW path           :", RAW_DATA_DIR)
print("Processed path     :", OUTPUT_DATASET)
```

### Importer les classes

```python
# ---------------------------------------------------------
pre = TextPreprocessor(min_length=40, max_length=400)
builder = DatasetBuilder()
```

### Lecture + découpage des documents RAW


```python
all_chunks = []

for filename in os.listdir(RAW_DATA_DIR):
    if filename.endswith(".txt"):
        filepath = os.path.join(RAW_DATA_DIR, filename)
        print("Processing:", filepath)

        with open(filepath, "r", encoding="utf-8") as f:
            text = f.read()

        chunks = pre.split_into_chunks(text)
        all_chunks.extend(chunks)

print(f"\n✔ Nombre total de chunks : {len(all_chunks)}")
```

### Sauvegarde du dataset

```python
builder.save_dataset(all_chunks, OUTPUT_DATASET)

print("✔ Dataset créé :", OUTPUT_DATASET)
```
